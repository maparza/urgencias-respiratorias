# Machine Learning
## Creación de variables temporales

```python
# Ordenar cronológicamente por región
df_modelo = df_semanal.sort_values(
    ["RegionGlosa", "Anio", "SemanaEstadistica"]
).copy()

df_modelo.head()
```

```python
# Atenciones de semanas anteriores
df_modelo["lag_1"] = (
    df_modelo.groupby("RegionGlosa")["NumTotal"].shift(1)
)

df_modelo["lag_2"] = (
    df_modelo.groupby("RegionGlosa")["NumTotal"].shift(2)
)

df_modelo["lag_3"] = (
    df_modelo.groupby("RegionGlosa")["NumTotal"].shift(3)
)
```

```python
df_modelo["promedio_3_semanas"] = (
    df_modelo[["lag_1", "lag_2", "lag_3"]]
    .mean(axis=1)
)
```

```python
#calcular el percentil 75 por región
umbrales = (
    df_modelo.groupby("RegionGlosa")["NumTotal"]
    .quantile(0.75)
)

umbrales
```

```python
#umbral para cada registro
df_modelo["umbral_alta_demanda"] = (
    df_modelo["RegionGlosa"].map(umbrales)
)
```

```python
#clasificar siguiente semana; no actual
df_modelo["NumTotal_siguiente"] = (
    df_modelo.groupby("RegionGlosa")["NumTotal"].shift(-1)
)
```

```python
#definición de demanda
df_modelo["AltaDemanda"] = (
    df_modelo["NumTotal_siguiente"]
    >= df_modelo["umbral_alta_demanda"]
).astype(int)
```

```python
#remover blancos comportamientos anteriores
df_modelo = df_modelo.dropna(
    subset=[
        "lag_1",
        "lag_2",
        "lag_3",
        "NumTotal_siguiente"
    ]
).copy()
```

## X e Y

```python
features = [
    "SemanaEstadistica",
    "lag_1",
    "lag_2",
    "lag_3",
    "promedio_3_semanas"
]

X = df_modelo[features]
y = df_modelo["AltaDemanda"]

print(X.shape)
print(y.shape)
```

## train & test

```python
train = df_modelo[df_modelo["Anio"] <= 2024]
test = df_modelo[df_modelo["Anio"] >= 2025]
```

```python
X_train = train[features]
y_train = train["AltaDemanda"]

X_test = test[features]
y_test = test["AltaDemanda"]

print(X_train.shape)
print(X_test.shape)
```

## estandarización

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## regresión logística

```python
from sklearn.linear_model import LogisticRegression

modelo = LogisticRegression(
    max_iter=1000,
    class_weight="balanced"
)

modelo.fit(X_train_scaled, y_train)
```

## predicción

```python
y_pred = modelo.predict(X_test_scaled)

y_prob = modelo.predict_proba(X_test_scaled)[:, 1]
```

```python
y_prob
```

## Evaluación

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)

print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nROC-AUC:")
print(roc_auc_score(y_test, y_prob))
```

# Modelo final 

```python
features_finales = [
    "SemanaEstadistica",
    "RegionGlosa",
    "lag_1",
    "lag_2",
    "lag_3",
    "promedio_3_semanas"
]
```

```python
X_train = train[features_finales]
y_train = train["AltaDemanda"]

X_test = test[features_finales]
y_test = test["AltaDemanda"]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
variables_numericas = [
    "SemanaEstadistica",
    "lag_1",
    "lag_2",
    "lag_3",
    "promedio_3_semanas"
]

variables_categoricas = [
    "RegionGlosa"
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        (
            "num",
            StandardScaler(),
            variables_numericas
        ),
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            variables_categoricas
        )
    ]
)
```

```python
modelo_final = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "classifier",
            LogisticRegression(
                max_iter=1000,
                class_weight="balanced"
            )
        )
    ]
)
```

```python
modelo_final.fit(X_train, y_train)
```

```python
y_pred_final = modelo_final.predict(X_test)

y_prob_final = modelo_final.predict_proba(X_test)[:, 1]
```

```python
print("Accuracy:", accuracy_score(y_test, y_pred_final))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_final))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_final))

print("\nROC-AUC:")
print(roc_auc_score(y_test, y_prob_final))
```

## Exportación del modelo

```python
import pickle

with open("modelo.pkl", "wb") as archivo:
    pickle.dump(modelo_final, archivo)

print("Modelo guardado correctamente")
```
